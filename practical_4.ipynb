```python
import time

# Iterative factorial
def factorial_iterative(n):
    fact = 1
    for i in range(1, n + 1):
        fact *= i
    return fact


# Recursive factorial
def factorial_recursive(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial_recursive(n - 1)


# Input
n = int(input("Enter a number: "))

if n < 0:
    print("Factorial is not defined for negative numbers.")
else:
    # Iterative method
    start = time.perf_counter()
    result_iterative = factorial_iterative(n)
    end = time.perf_counter()
    iterative_time = end - start

    # Recursive method
    start = time.perf_counter()
    result_recursive = factorial_recursive(n)
    end = time.perf_counter()
    recursive_time = end - start

    # Output
    print("\nFactorial using Iterative Method:", result_iterative)
    print("Time taken:", iterative_time, "seconds")

    print("\nFactorial using Recursive Method:", result_recursive)
    print("Time taken:", recursive_time, "seconds")

    print("\nTime Complexity:")
    print("Iterative: O(n)")
    print("Recursive: O(n)")

    print("\nSpace Complexity:")
    print("Iterative: O(1)")
    print("Recursive: O(n)")
```
